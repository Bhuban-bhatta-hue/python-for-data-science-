```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
#task1
wekip=requests.get('https://www.worldometers.info/world-population/population-by-country/')
print(wekip)
soaps=BeautifulSoup(wekip.content,'lxml')
print(soaps)
table2=soaps.find("table",class_="datatable w-full border border-zinc-200 datatable-table")
print(table2)
```

```python
web=requests.get('https://www.iplt20.com/auction/2022')
print(web)
soup=BeautifulSoup(web.content,'lxml')
print(soup)
```

```python
soup.prettify()
```

```python
table1=soup.find("table",class_="ih-td-tab w-100 auction-tbl")
print(table1)
```

```python
columns=table1.find_all('th')
for i in columns:
  print(i.text)
```

```python
table_body=table1.find("tbody",id="pointsdata")
print(table_body)
```
