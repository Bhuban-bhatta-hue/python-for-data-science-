```python
#habdling categorical values
df=pd.get_dummies(df,columns=cat_cols)
```

```python
print(df.head)
```

```python
x=df.drop('Price',axis=1)
y=df['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.3,random_state=22)
```

```python
x.shape
```

```python
y.shape
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
#train the model
from sklearn.linear_model import(
    LinearRegression,
    Ridge,
    Lasso,
    ElasticNet,
    LogisticRegression
    )
model1=LinearRegression()
model2=Ridge()
model3=Lasso()
model4=ElasticNet()
model5=LogisticRegression()
```

```python
model1
```

```python
model1.fit(x_train,y_train)
```

```python
print('the slop or coefficient of weight is:',model1.coef_)
print('the intercept of the line is:',model1.intercept_)
```

```python
y_pred_test=model1.predict(x_test)
```

```python
y_pred_test,y_test
```

Performance metrics MSE=Mean square error RMSE=root mean square error MAE=mean accelerator error

R2 score,Adjusted R2 Score

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
mse=mean_squared_error(y_test,y_pred_test)
mae=mean_absolute_error(y_test,y_pred_test)
rmse=np.sqrt(mse)
print("mean square error:",mse)
print('mean absolute error:',mae)
print("root mean aquared error:",rmse)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred_test)
print(score)
print("r2 score:",score*100,'%')
```

```python
x_test.shape
```

```python
x_test.shape[1]
```

```python
#display adjustinf r2 score
adjusted_r2=1-(1-score)*(len(y_test)-1)/(len(y_test)-x_test.shape[1]-1)
print('adjusted r2 score:', adjusted_r2*100,'%')
```

```python
# polynomial regression

## Applying Polynomial Transformation
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree = 2)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
```

```python
x_train_poly
```

```python
from sklearn.metrics import r2_score
lin_reg2 = LinearRegression()
lin_reg2.fit(x_train_poly, y_train)
y_pred = lin_reg2.predict(x_test_poly)
score = r2_score(y_test, y_pred)
print('R2 score for Polynomial Regression:', score)
```

```python
print('Coefficients:', lin_reg2.coef_)
```
